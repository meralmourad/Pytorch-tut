```python
import torch 
from torch import nn
import matplotlib.pyplot as plt

Divice = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
# linear regression formula is Y = weight * X + bias
weight  = 0.7 
bias = 0.3

start = 0
end = 1
step = 0.02

X = torch.arange(start , end , step).unsqueeze(dim=1)
Y = weight * X + bias 

print(X[:10],'\n\n', Y[:10],'\n\n', 'So the Data is', len(X), 
    'samples\nAnd thier actual values are Y' )
```

```python
train_split = int(0.8 * len(X))
X_train , y_train = X[:train_split],Y[:train_split]
X_test , y_test = X[train_split:],Y[train_split:]

print(len(X_train),len(y_train),len(X_test),len(y_test))
```

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(1, #seed 
                                                requires_grad=True ,
                                                dtype=torch.float))
        self.bias = nn.Parameter(torch.randn(1 , 
                                                requires_grad=True ,
                                                dtype=torch.float))

        # override
    def forward(self , x: torch.Tensor) -> torch.Tensor:
        return self.weight * x + self.bias # linear regression formula
```

```python
torch.manual_seed(42)
model_0 = LinearRegressionModel()
print(list(model_0.parameters()))
```

```python
with torch.inference_mode():
    y_preds = model_0(X_test)

with torch.no_grad():
    y_preds = model_0(X_test)

y_preds , y_test
```
